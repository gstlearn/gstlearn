## Import packages

```python
import numpy as np
import gstlearn as gl
import gstlearn.plot as gp
import gstlearn.document as gdoc
```

Global variables

```python
gl.OptCst.define(gl.ECst.NTCOL, 6)
gdoc.setNoScroll()
```

## Reading data

The data are stored in a CSV format in the file called Pollution.dat 

```python
filepath = gdoc.loadData("Pollution", "Pollution.dat")
mydb = gl.Db.createFromCSV(filepath, gl.CSVformat())
mydb.setLocators(["X", "Y"], gl.ELoc.X)
mydb.setLocator("Zn", gl.ELoc.Z)

dbfmt = gl.DbStringFormat()
dbfmt.setFlags(flag_resume=True, flag_vars=True, flag_extend=True)
mydb.display(dbfmt)
```

Accessing to the variable names

```python
print("List of all variable names =", mydb.getAllNames())
```

Extracting the vector containing the Zn variable in order to perform a selection

```python
mydb.setLocator("Zn", gl.ELoc.Z)
mydb.getStatsAsTable().display()
```

Display my Data (with samples represented by color and size). Note the presence of three outliers: they are kept in this workflow for the demonstration of some features of the Variogram Cloud.

```python
gp.plot(mydb, nameSize="Zn", nameColor="Zn")
gp.decoration(title="Data Set")
```

## Variogram Cloud

```python
myVarioParamOmni = gl.VarioParam()
mydir = gl.DirParam.create(nlag=10, dlag=1.0)
myVarioParamOmni.addDir(mydir)
```

We calculate the Variogral Cloud

```python
vcloud = gl.vcloudFromDb(mydb)
```

We recall that the Variogram cloud is calculated by filling an underlying grid where each cell is painted according to the number of pairs at the given distance and given variability. Representing the variogram cloud

```python
gp.plot(vcloud, "Cloud*")
gp.decoration(title="Variogram Cloud")
```

We calculate the variance of the variable of interest (e.g. Zn) adn compare it with the vertical extension of the variogram cloud.

```python
expvar = mydb.getVariance(mydb.getNameByLocator(gl.ELoc.Z))
print("Experimental Variance =", expvar)
```

We want to identify the samples which are responsible for the pairs with high variability (e.g. above the experimental variance), located at short distance (say below 10).

```python
vcloud = gl.vcloudFromDb(mydb, distmax=15, varmin=expvar, countmax=5)
```

We clearly see that the samples with ranks 74, 31 and 102 (1-based) could be considered as outliers. 

We wish to demonstrate this by highlighting the impact of the first two samples in the Variogram Cloud calculations.

Let us start with sample ranked #74 (e.g. #73 in 0-based numbering). We calculate the variogram cloud obtained when comparing it to all samples of the Db. We overlay this partial variogram cloud to the initial global one.

```python
Rank = 74
vcloud1 = gl.vcloudFromDb(mydb, samples=[Rank - 1])
gp.raster(vcloud)
ax = gp.symbol(vcloud1, nameSize="Cloud.Zn", flagCst=True, c="blue", marker="*", s=100)
```

```python
Rank = 31
vcloud1 = gl.vcloudFromDb(mydb, samples=[Rank - 1])
gp.raster(vcloud)
ax = gp.symbol(vcloud1, nameSize="Cloud.Zn", flagCst=True, c="blue", marker="*", s=100)
```
