```python
import gstlearn as gl
from IPython.display import display, Latex
```

This script is meant to demonstrate the use of the LateX interface functions.

### Dense matrices

A Dense matrix is generated with random values

```python
neq = 5
matrix = gl.MatrixDense.createFillRandom(neq, neq)
matrix.display()
```

```python
display(Latex(matrix.toLatex()))
```

### Sparse matrices

A Sparse matrix is generated with random values

```python
neq = 5
matrix = gl.MatrixSparse.createFillRandom(neq, neq, 0.2)
matrix.display()
```

```python
display(Latex(matrix.toLatex()))
```

### Tables

A Table is generated by storing the statistics calculated on an input Db.

```python
mydb = gl.Db.createFillRandom(100, 2, 3)
table = mydb.displayStats()
table.display()
```

```python
display(Latex(table.toLatex()))
```
